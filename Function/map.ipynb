## Map()


Map function applies a given function operation to alll items in an iterable or a List and returns a map object(An Iterator)

Map() is going to help us out with avoiding unnecessary lines of code and we can apply operation on all values or all items in an iterable and this is the sign of a good programmer.

### Syntax for Map()

### list(map(function,iterable))

```python
## Squaring all items of a list

def square(list_items):
    empty_list=[]
    for items in list_items:
        squared_item=items**2
        empty_list.append(squared_item)
    print(list_items)    
```

```python
square([1,4,5,3,1,8])
```

```python
## similiar problem statement with a Map()-->

print(list (map(lambda x:x**2, [1,2,4,5,6,9])))
```

```python
### Map Function with basic function in it

def Squared_numbers(num):
    return num**2
print(list(map(Squared_numbers,[1,2,4,5,6,9])))
```

## Whenever you are going to pass the general function inside your map function as the parameter then you dont have to use parenthesis.    

## While Mapping the value of the one iterable with another iterable we can actually apply the operation in between the values of the two iterable in a synchronous way

[1,2,3,4]<---> [5,6,8,2]---->Addition --->[6,8,11,6]

```python
## Map function with mulitple iterables
numbers_1=[1,2,3,4]
numbers_2=[5,6,8,2]

print(list(map(lambda x,y:x+y,numbers_1,numbers_2)))
```

```python
### Map() to convert all String values items in an iterable into a integer

map_values=list(map(int,['1','3','5','7','9']))
```

```python
print(map_values)
```

```python
employes_names=['amit','shikha','vaibahv','shubh']

print(list(map(str.upper,employes_names)))
```

```python
### Retiever the values inside a dictionary using map()

employees=[{'Name':'Yash','Age':23,'Qualificaton':'Btech'},
           {'Nmae':'Amit','Age':24,'Qualification':'Btech'}]
```

```python
def age_retiever(list_items):
    return list_items['Age']
```

```python
print(list(map(age_retiever,employees)))
```
