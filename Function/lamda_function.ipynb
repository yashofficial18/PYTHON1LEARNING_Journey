## Lambda Function

### Lambda Function are the one Liner Function Definrd by the Lambda keyword

## syntax for Lambda Function:

## lambda arguments:expression

## lambda --> kayword which simpally let the python interpreter know that the current function is our lambda function 

## Arguments -->Input values for our Function 

## Expression ---> Logic (Function Body )

#### lambda x:x**2

## They can have any number of arguments but only one expression or only one Logic
## They are used for shoot Operations or as arguments to higher Order Functions
## It can be treated as a Function without a Name and they Dont even require a Return Statement 

```python
## fuction which calculates the square of the numbers 
def square(num):
    return num**2
```

```python
square(5)
```

```python
## now we will do the same thing by using lambda function 

lambda_function=lambda x:x**2
```

```python
type(lambda_function)
```

```python
lambda_function(5)
```

```python
## lambda function to calculate the sum of two numbers
sum=lambda x,y:x+y
print(sum(5,6))
```

```python
product=lambda a,x,y,z:a*x*y*z
print(product(2,9,4,6))
```
