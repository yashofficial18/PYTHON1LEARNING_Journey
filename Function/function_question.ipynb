```python
## To create function that converts temperature from celcius to Fahrenheit or Fahrenheit to Celsius

## if temperature is in celcius, then celsius to Fahrenheit conversion is :
## 1 deg celsius= 9/5 +32

def converts_temperature(temp_value,unit):
    if unit=="C":
        temperature=temp_value*9/5+32
        print(f"The temperature for {temp_value} deg Celsius in Fahrenheit is: ", str(temperature) + "F")
    elif unit=="F":
        temperature=(temp_value-32)*9/5
        print(f"The temperature for {temp_value} deg Fahrenheit in Celsius is: ", str(temperature) + "C") 
    else:
        print("Invalid Value")

```

```python
temp_value=int(input("Enter the Temperture value:"))
unit=str(input("Enter the unit:"))
converts_temperature(temp_value,unit)
```

```python
## define a function thta would behave as a password sthrength checker
'''
strong password -->
if your password:
1. length is greater than 8 characters
2. password must consist of atleast a lowercase character (a,b,c...z)
3. password must consist of atleast a uppercase character (A,B,C...Z)
4. password must consist of atleast a digit (1,2,)
5. password must consist of atleast a special character (!,@,#,$,%)
'''
## weak password-->
'''
    if a password doesnt match any above condition then it will be considered as a weak password  
'''
```

```python
### islower()--> checks that wheather if the current character is a lowercase character or not

### isupper()--> checks that wheather if the current character is a uppercase character or not

## isdigit()-->checks that wheather if the current character is a digit or not
```

```python
def password_strength(password):
    if len(password)<8:
        print(f"The Entered {password} password is weak.")

    elif not any (char.islower() for char in password):
        print(f"The Entered {password} password is weak.")

    elif not any (char.isupper() for char in password):
        print(f"The entered {password} password is weak.")

    elif not any (char.isdigit() for char in password):
        print(f"The Entered {password} password is weak.")

    elif not any (char for char in '!,@,#,$,%,^,&,*,(,),_,+,'):
        print(f"The Entered {password} password is weak.")  
             
    else:
        print(f"The Entered {password} password is Strong.")    

```

```python
password=str(input("Enter your Password:"))
password_strength(password)
```

```python
## palindrome string 
## palindrome string is going to have character similar to it from both the sides or even if you will reverse the 
## string still you will be getting the given string itself

## palidrome --> aba, aaaabbaa , zzyzz
## not a palindrome --> Ab,  Aba, abA, abaa
```

```python
## check that wheather if the current string is a palindrome or not

def palindrome(string):

    string_value=string.lower().replace(" ", "")

    if string_value==string_value[::-1]:
        print("the current string is palindrome")

    else:
        print("the current string is not palindrome")     
```

```python
string=str(input("Enter the character:"))
palindrome(string)
```
