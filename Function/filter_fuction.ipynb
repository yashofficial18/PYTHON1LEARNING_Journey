## Filter Function

## Filter Function is going to retieve only those items which are meeting a certain condition in an iterable 

### The filter function construct an iterator from elements of an iterable for which a condition remains True or for whicha function returns True Value   

## Syntax 

## list(filter(func,iterable))

```python
## to reteive only those values which are even 

def even_function(value):
    if value%2==0:
        return True
```

```python
retiever_list=(1,2,3,4,565,7,78,90,55,6,3,665,3,5,33,43,56,78) 

print(list(filter(even_function,retiever_list)))
```

```python
## same kind of problem solving by lambda function

print(list(filter(lambda x:x%2==0,retiever_list)))
```

```python
##  I Want to retiever the values which are greater than 5
retiever_greater=(1,2,3,4,5,6,7,8,9,)
print(list(filter(lambda value:value>5,retiever_greater)))
```

```python
# if ui want to retieve all those values which are greater than 5 as well and are even as well

list_items=(2,3,4,0,1,12,5,67,90,89,101,10)  
print(list(filter(lambda x:x>5 and x%2==0,list_items)))
```

```python
employees= [{'Name':'Shubh','Age':28,'Qualification':'BA'},
            {'Name':'Vaibhav','Age':27,'Qualification':'Ai'},
            {'Name':'Amit','Age':25,'Qualification':'Btech'}]
```

```python
def retieve_age(employe):
    if employe['Age']>25:
        return True
```

```python
print(list(filter(retieve_age,employees)))
```
