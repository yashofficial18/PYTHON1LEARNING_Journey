## Recursive function--> By a Recursive function we can say that a function is going to call itself again and again

```python
## Factorial of a Number
## Factorial of 8--> 8*7*6*5*4*3*2*1
## Factorial of 0 or Factorial of 1 ---> 1 
```

```python
def Factorial(num):
    if num==0 or num==1:
        print(f" The factorial of {num} is: 1")

    else:
        Factorial=1
        for i in range(1,num+1):
            Factorial=Factorial*i
        print(f"The Factorial of {num} is {Factorial}")    
```

```python
num=int(input("Enter the integer:"))
Factorial(num)
```

```python
## Method 2
def Factorial_recursive(num):
    if num==0 or num==1:
        return 1
    else:
        return num*Factorial_recursive(num-1)  
          
Factorial_recursive(5)
```

```python
## return num*Factorial_recursive(num-1)  
'''
base condition ---> which will stop the loop or which will let the function to stop calling itself in our base,
base condition was ---> i if num==0 or num===1 -->1
return num*Factorial_recursive(num-1) 
num==5 
5*factorial_recursive(4) & factorial_recursive(4) ---> 4*factorial_recursive(3)
factorial_recursive(3)--> 3*factorial_recursive(2)
factorial_recursive(2)--> 2*factorial_recursive(1)
'''
```

```python
input_recursive=int(input("entr the integer"))
print(f"the factorial of {input_recursive} is:",Factorial_recursive(input_recursive))
```

## Positional Arguments--> Position arguments are being used so that we can pass multiple values inside our function as an argument irrespective of the naumber of values we have mentuoined as the parameter inside our function   


```python
# function to add multiple numbers irrespective of how many numbers we are passing 

def addition_position(*num):
        add=0
        for i in num:
            add+=i
        print("The addition of the number are ",add)
```

```python
addition_position(1,2,3,4,5)
```

## Keyword Arguments --> takes the input as the keyword values and irrespective of any number of keyword values you are going to pass it wont matter

```python
def data(**user_data):
    for key, value in user_data.items():
        print(f'{key}:{value}')
```

```python
data(name="Amit", Sex="Male", Age=27)
```

## you guys want to pass keyword value pairs as your input along with postional arguments as well

```python
def variable(*positional,**keyword):
    for i in positional:
        print(f"postional values are :",i)
    for key,value in keyword.items():
        print(f"key is: {key} and value is: {value}")
```

```python
variable(1,2,3,4,5, name="amit", age=27, gender="Male")
```
