## Polymorphism


## Polymorphism means Many forms in greek.
## In OOPS, Polymorphism allows object of Different classes as if they are the object of the same class & It enables reusability and makes programs more flexible.

## Types OF Polymorphism:
1. Method Overriding (Rutime Polymorphism)
2. Method Overloading (Compile Time Polymorphism)
3. Operator Overloading 

Method Overriding -->  A child class provides a specific implementation of a method that has already been defined into the parent class.
This allows Dynamic Selection at Runtime.

```python
class Grandparent():
    def speak(self):
        print("Hello! My name is Amit")

class Parent():
    def speak(self):
        print("Hello! My name is Junior Amit")

class Child():
    def speak(self):
        print("Hello! My name is Junior Junior Amit")
```

```python
def attribute_provider(objects):
    print(objects.speak())
```

```python
grandpa=Grandparent()
```

```python
attribute_provider(grandpa)
```

```python
child=Child()
```

```python
attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "The method returns the area of the shape"
```

```python
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    
    def area(self):
        return self.width*self.height

class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius

```

```python
shape=Shape()
shape.area()
```

```python
rectange=Rectangle(6,4)
rectange.area()
```
