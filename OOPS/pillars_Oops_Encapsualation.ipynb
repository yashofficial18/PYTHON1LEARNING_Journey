# Encapsulation

## Encapsulation is the way of wrappping data (Variables and Methods) inside a capsule or a class as a single unit.

## It restricts the direct access to some of the objects components which is a means of preventing accidental inference and misuse of the data.

## Key concepts of Encapsulation:
## 1. Public Members 
## 2. Protected Members
## 3. Private Members
## 4. Getter and Setter methods

## Punlic Members---> Acessible from Anywhere or Attributes are accessible outside of the class and these members are accessible by the derived class

## public member attribute is directly declared 

```python
class Ironman:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def introduction(self):
        print(f"My name is {self.name} and i have {self.weapon}")
```

```python
## dir()--> it takes the input as the object name and gives us this idea that which member is public or protected or private. or it will show you all the attritube regading that particular class
```

```python
iron=Ironman("Ironman","Genius Intellect","Repulsor Beam")
```

## Private Members---> A private member is not accessible outside the class and not even in the derived class and in order to have the access over these private members or to bring some modifiaction on thes members we use getter and setter method.

## Private Members are declared by using double underscore __before the name 

```python
class Avengers:
    def __init__(self,name,phone_number):
        self.name=name
        self.__phone_number=phone_number     ##private
    
    def introduction(self):
        print("my name is {self.name}")
```

```python
black=Avengers("Iron","384389")
```

```python
black.name
```

```python
black.phone_number
```

```python
## Getter Function

## Getter method---> it will be helping us out with accessing the private data
```

```python
class Captian:
    def __init__(self,name,address):
        self.name=name
        self.__address=address
    
    def accessing_add(self):
        print(f"My address is {self.__address}")
```

```python
captain1=Captian("Captain America","Newyork")
```

```python
captain1.name
```

```python
captain1.address
```

```python
captain1.accessing_add()
```

```python
#Setter Method---> Settter Method is being used for modifying our private data and 
```

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound
        
    def sound_info(self):     #getter method
        print(f"The Aniaml is going to sound as {self.__sound}")
    
    def sound_modify(self,new_sound):
        print(f"The new sound of the Animal is {new_sound}")
        
```

```python
dog=Animal("Bruno","Bhou bhou")
```

```python
dog.name
```

```python
dog.__sound
```

```python
dog.sound_info()
```

```python
dog.sound_modify("meow meow")
```

## Protected data is the data which is not accessible outside the class but the protected data is going to be accessible by the derived class or child class

## For having the protected member you can mention your protected member as '_' attribute or by using a single underscore

```python
class Grandparent:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    def locker_info(self):
        print(f"The locker Key info is {self._locker_key}")
```

```python
class Parent(Grandparent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"The locker key information is: {self._locker_key}")
```

```python
grandpa=Grandparent("Amit",12345)
```

```python
grandpa.name
```

```python
grandpa._locker_key
```

```python
parent=Parent("Amit",12345)
```

```python
parent.name
```

```python
parent._locker_key
```

```python
parent.child_reveals()
```

```python
class Person:
  def __init__(self, name, salary):
    self.name = name
    self._salary = salary # Protected property

p1 = Person("Linus", 50000)
print(p1.name)
print(p1._salary) # Can access, but shouldn't
```
