## Abstraction

## Abstraction is the concept of  hiding the implementation details and only showing the necessary features of the object.

## This helps in reducing programming complexity and efforts 

## Abstraction alllows us to work with an object with what an object does rather than how it does it 

### How can we achieve abstraction in python

## 1. Abstract class ---> A class that cannot be instantiated or that means for an abstract class we cant create the object of this class.

## 2. Abstract Method ---> Abstract method just because it cant accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class.

## The abstract method is being defined by 
@abstractmethod

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    def stop(self):
        pass
    
class Car(Vehicle):

    def start(self):
        print("The Car Start with ignite")

    def stop(self):
        print("The car stop with brake peddle")

class ElectricScooter(Vehicle):

    def start(self):
        print("The Electric Scooter Start with Electric Button")

    def stop(self):
        print("The Electric Scooter Stop With Electric Brake")
```

```python
car=Car()
```

```python
car.start()
```
