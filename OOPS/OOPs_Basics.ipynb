```python
class Avenger:
    def __init__(self,name,power,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"Hello! My name is {self.name}")

    def carrying_weapon(self):
        print(f"I am carrying the weapon {self.weapon}")
```

```python
Iron_man=Avenger("Iron man","Genius Intellect and Technology","Repulsor beans")

print(Iron_man.introduction())
print(Iron_man.carrying_weapon())
```

## Types of Method

## 1. Instance Method : Instance Method operate on Object Attributes.
## The Instance Method is going to use self keyword to access and modify object attributes.
## These Methods are called using an object.


## 2. Class Method : It is the method of the whole class  or the whole blueprint which will remainthe same for all of the object then such kind of Method is being reffered as a class Method.
## In Class Method , These Method work with the class level attributes.
## They are going to use class instead of self
## For Specific that the current method is a class method We have to use abbreviation as @classmethod  


## 3. Static Method : Static Method do not use self or cls.
## They work like normal functions but belongs to a class.
## They are going to be designed by a abbreviation @staticmethod 
## static method can also be called by the classname also 

```python
## instave method Example 

class Avengers:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    
    def show_details(self):
        print(f"My name is {self.name} and i have {self.power} power")
        
```

```python
iron_man=Avengers("Iron Man", "genius Intellect")
```

```python
iron_man.show_details()
```

```python
thor=Avengers("Thor","God of Thunder")
```

```python
thor.show_details()
```

```python
## class Example

class Avengerss:

    origin="Avenger"
    
    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")
    
    def show_details(self):
        print(f"My name is {self.name} and i have {self.power} power")
```

```python
iron_man=Avengerss("Iron Man", "genius Intellect")
```

```python
iron_man.name
```

```python
iron_man.power
```

```python
iron_man.show_details()
```

```python
iron_man.origin_info()
```

```python
thor=Avengerss("Thor","God of Thunder")
```

```python
thor.origin_info()
```

```python
## STATIC METHOD 

class Avengersss:

    @staticmethod
    def mission():
        print("All Avengers must Save the earth")

    origin="Avenger"
    
    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")
    
    def show_details(self):
        print(f"My name is {self.name} and i have {self.power} power")
```

```python
iron_man=Avengersss("Iron Man", "genius Intellect")
```

```python
iron_man.mission()
```

```python
Avengersss.mission()
```
