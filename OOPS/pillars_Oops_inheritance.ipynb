## Pillars of Object Oriented Programming

## Inheritance 

### Inheritance  is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the  properties Or Functionalities of the parent class

### just like A child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definitely in this hierarchy the grandchild can have some features differnt from the parent and grandparent as well 

#### In python Typically we got to see  3 Types of inheritance:

#### 1. Single inheritance 
#### 2. Multilevel Inheritance 
#### 3. Multiple Inheritance

#### Single Inheritance ----> The Feature or the Attributes from one parent class will be getting inherited into one child class.

#### Multilevel Inheritance----> The Properties or methods and attributes of the grand parent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class. In a proper heirarchy the Inheritance process is going to work for us.

```python
class Avenger:
    
    def __init__(self,name,power):
        self.name=name
        self.power=power
    
    def introduction(self):
        print(f"Hello My name is {self.name}")

class Ironman(Avenger):

    def __init__(self,name,power,weapon):
        super().__init__(name,power)
        self.weapon=weapon

    def weapon_details(self):
        print(f"I use {self.weapon}")
```

```python
class Grandparent:
    def __init__(self,nationality,baldness,anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def introduction(self):
        print(f"Hello My Nationality is : {self.nationality} and about my anger_issues if you want to ask about then the answer is : {self.anger_issues}")
       


```

```python
class Parent(Grandparent):

    def __init__(self,nationality,baldness,anger_issues,intelligence):
        super().__init__(nationality,baldness,anger_issues)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f" About my intelligence if you guys want to know that wheather if i am intelligent or not then the answer is : {self.intelligence}")
```

```python
class Child(Parent):
    def __init__(self,nationality,baldness,anger_issues,intelligence,charm):
        super().__init__(nationality,baldness,anger_issues,intelligence)
        self.charm=charm

    def charm_details(self):
        print(f"About my charm If you guys want to know that wheather if i am charming or not then the answer is : {self.charm}")
```

```python
grandparent_1=Grandparent("Indian","little bit","YES")
```

```python
parent_1=Parent("Indian","YES","NO", "YES")
```

```python
child_1=Child("Indian","NO","YES","Yes","Yes")
```

```python
class Ironmans:

    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality.nationality

    def introduction(self):
        print(f"My name is {self.name} and power are {self.power}")


class Captainamerica:
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality.nationality

    def weapon_details(self):
        print(f"I carry the weapon {self.weapon}")
```
