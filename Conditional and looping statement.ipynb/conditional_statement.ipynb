```python
#simple Calculator

a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))

operation = input("Enter the opertion out of the +, -, *, /, %")

if operation == "+":
    print("the sum of a and b is: ", a+b)

if operation == "-":
    print("the sub of a and b is: ", a-b)

if operation == "*":
    print("the sum of a and b is: ", a*b)

if operation == "/":
    print("the sum of a and b is: ", a/b)
    
```

```python
year=int(input("Enter the year Number: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print("The current year is leap year.")
        else:
            print("The current year is not leap year.")
    else:
        print("year is leap year")        
else:
    print("year is not a leap year")                           
```
