## open()---> opening the file 
## First Parameter ---> File_name
## Second Parameter ---> Mode
##  Mode ---> it decides the way in which we are working with this file

 ### Mode---> r ---> Reading the content from the file
 ### Mode---> w ---> Write the content into the file
 ### Mode---> r+ ---> Read the content first and then writing the content into the file
 ### Mode---> w+ --->Writing the content first and then reading from it

### Reading your File
### read()---> Reads the content from the File

```python
f=open('example.txt','r')
content=f.read()
print(content)
f.close()
```

## Writing the content into the file 
### write()---> helps us out in writing the content into the file 
### append()--->  it adds the data to the end of the file without overwriting the existing data in the file

```python
f=open('example.txt','w')
f.write('This is an example of file handling in python')
f.close()   
```

```python
## appending th econtnet into the our file 

f=open('example.txt','a')
f.write('\n We are learning about Gen AI Pre requisite Python')
f.close()
```

## With ---> It willbe used for first of all opening the file as a object and there is no requirement for closing the file as well   

```python
with open('example.txt','r') as f:
    content_data=f.read()
    print(content_data)
```

```python
with open('example.txt','w') as file:
    file.write('i have learned the  with command')
    
```

```python
with open('example.txt','a') as files:
    files.write('\n we are exploring appending the content into the file using with command')
    
```

```python
## write the content into our file line by line ---> writelines() method 
```

```python
lines=['First Line\n', 'Second Line\n', 'Third Line\n']
with open('example.txt','w') as data:
    data.writelines(lines)  
```

```python
## Reading the Total Lines, Total Words, Total Characters from the file-->readlines()

def count_lines(file):
    with open(file,'r') as files:
        lines=files.readlines()
        total_lines=len(lines)
        total_words=sum(len(line.split()) for line in lines)
        total_characters=sum(len(line) for line in lines)
    return total_lines, total_words, total_characters

file_path='Total_data.txt'
Total_Lines,Total_Words,Total_Characters=count_lines(file_path)
print(f"Total Lines are : {Total_Lines} and Total Words are : {Total_Words} and Total Characters are : {Total_Characters}")
```

```python
with open('Total_data>txt','w+') as write_read_file:
    write_read_file.write("\n we are looking into W+ opertaion")
    write_read_file.write("\n it means writing and then reading")
    write_read_file.seek(0)
    content_complete_data=write_read_file.read()
    print(content_complete_data)
```
