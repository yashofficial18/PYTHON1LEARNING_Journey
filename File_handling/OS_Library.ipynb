##  OS Library is able to interact with our Operating System and can let us work with Mulyiple Directories and Files

```python
## getcwd()--> it is capable of providing us with the idea about the current directory 
## getcwd()--> it stand for Get Current Working Directory
```

```python
import os
current_path=os.getcwd()
print("The current working directoryis: ", current_path)
```

```python
## mkdir()---> it is stand for Make Directory ---> it will helping us out with creating the directory 
```

```python
new_path="New Folder"
os.mkdir(new_path)
print(f"The new directory {new_path} is created successfully")
```

```python
## listdir()---> it stand for list directory ---> Listing all of the directories present
All_items=os.listdir('.')
print(All_items)
```

```python
## joining multiple Paths Together
## os.path.join()---> it will be joining multiple path together

dir_name="Folder_Data"
file_name="File_Data"
Full_Path=os.path.join(dir_name,file_name)
print(f"The Full Path is: {Full_Path}")
```

```python
## I want to join Multiple Path with the current path

current_complete_path=os.path.join(os.getcwd(),dir_name,file_name)
print(f"The Complete path is : {current_complete_path}")
```

```python
## os path.exists()---> it will be checking the existence of the path

given_path="/Users/yashdwivedi/Desktop/PYTHON1/app.py" 

if os.path.exists(given_path):
    print(f"The given path {given_path} is Present")
else:
    print(f"THE given path {given_path} is not Exists")
```

```python
## os.path.isfile()---> This checks that whether if the current path is a file or not
## os.path.isdir()---> This checks that wheather if the current path is a directory or not

path='Total_data.txt'
if os.path.isfile(path):
    print(f"The given path {path} is a file")
else:
    print(f"The given path {path} is not a file")
```

```python
dr_path='destination.txt'
if os.path.isdir(dr_path):
    print(f"The given path {dr_path} is a directory")
else:
    print(f"The given path {dr_path} is a directory")
```

```python
path='New Folder'
if os.path.isfile(path):
    print(f"The given path {path} is a file")
elif os.path.isdir(path):
    print(f"The given path {path} is a directory")
else:
    print(f"The given path {path} is neither a file or not directory")
```

```python
## The Complete Path---> Absoulute Path---> os.path.abspath()---> It is the complete path of the file or directory
## os.path.abspath()--->It provides is with the complete path of the absolute path of the file or directory  

file_name='Total_data.txt'
complete_path=os.path.abspath(file_name)

print("The complete path or absolute path for the given file is", complete_path)
```
