```python
## Csv Library ---> Most of the time we are going to work with csv files so thats why knowledge with csv library fuctionality is very important.
## It is used to read and write csv files.  
```

```python
import csv
```

```python
with open('example.txt', mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Yash', '20', 'New York']) 
```

```python
with open('example.txt', mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## When we get the json Data, we get in a Json string format so by using json library we can convert our string data into dictionary
# format for serializing our data    
```

```python
import json
```

```python
data={"Name":"Amit", "Age":25}
json_str=json.dumps(data)
print(json_str)
print(type(json_str))
```

```python
json_dictionary=json.loads(json_str)
print(json_dictionary)
print(type(json_dictionary))
```

```python
## Regular Expression

## Regular Expresssion is used to retirieve info based on a specific pattern or if the pattern matches 
```

```python
import re 
text='Hello My name is amit and my password is HelloAmit12345'
```

```python
pattern=r'\d+'
match=re.search(pattern,text)
print(match.group())
```
